```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

data = [
    (1, "2024-01-01", "Noodles", 5, 10),
    (2, "2024-01-02", "Chocolate", 10, 5),
    (1, "2024-02-01", "Chips", 5, 10),
    (2, "2024-02-02", "Chocolate", 10, 5),
    (3, "2024-02-01", "Samosa", 5, 10),
    (1, "2024-03-01", "Chips", 5, 10),
    (4, "2024-03-02", "Chocolate", 10, 5),
    (5, "2024-04-01", "Samosa", 5, 10)
]

columns = ["customer_id", "date", "prod_name", "cost", "quantity"]

df = spark.createDataFrame(data, columns)

```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
w =  Window.partitionBy("customer_id").orderBy("date")
df_final = df.withColumn("rank", row_number().over(w)).filter(col("rank") == 1).withColumn("month", month(col("date"))).groupBy("month").count().orderBy("month")
display(df_final)
```
